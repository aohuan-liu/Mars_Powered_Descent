### 1.1 Solving Linear Systems 
###     求解线性方程组

```julia
# now let's solve some linear systems 
# A * x = b 
# 现在求解一些线性方程组： A*x = b 


x = inv(A)*b # you SHOULD NOT solve linear systems like this 
             # 你 ！！不应该！！ 这样解线性方程组

x = A\b      # you SHOULD solve them like this 
             # 应该这样解线性方程组
#=
 由于求逆矩阵:inv(A)属于O(n^3)的运算，复杂度高，
 直接用A\b做，反斜杠运算符会根据A矩阵的特性选择最合适的数值算法
=#


@test norm(A*x - b) < 1e-10 
#= 
由于有浮点数精度的限制，在数值计算中无法要求 A*x-b=0
1. A*x-b 称为残差(Residual)，把求解得到的解x代入原方程Ax=b后，等式两边的差值
2. norm(...)默认为2-范数(欧几里得范数)，把向量形式的残差转换成一个非负的标量，
   用来量化残差的大小，在2-范数中可以等价为求模
3. <1e-10，人为设定的误差阈值(Tolerance)，十亿分之一级别的误差
=#
```

```julia
# matrix factorizations (advanced usage) 矩阵分解(进阶用法)

# if I know something about the structure of my linear system, I can 
# choose how to factorize(分解) is 
# 如果我了解线性系统的结构，我就可以选择如何进行矩阵分解


# create a random positive definite matrix 
# 创建一个随机正定矩阵
A = randn(3,3); A = A' * A + I; 
# this just means it will work for all factorizations
# 这仅表示它适用于所有矩阵分解方法

# cholesky is the fastest factorization, but only works on positive definite matrices
# cholesky 是最快的分解方法，但仅适用于正定矩阵
chol_factor = cholesky(A) # factorize into L*L'=A, where L is lower triangular
                          # 分解为 L*L'=A，其中 L 为下三角矩阵
L = chol_factor.L  # this is the lower triangular cholesky factor 
                   # 这是下三角 cholesky 因子
@test norm(L*L' - A) < 1e-10

# some other popular options that work for all full rank square matrices: 
# 另一些适用于所有满秩方阵的常用方法
qr_factor = qr(A)
Q = qr_factor.Q 
R = qr_factor.R 
@test norm(Q*R - A) < 1e-10

# LU decomposition (LU分解)
lu_factor = lu(A)
L = lu_factor.L 
U = lu_factor.U 
# lu(A) 默认做行交换(pivoting)，实际分解是 PA = LU
# 所以验证用 P*L*U ≈ A，而不是 L*U ≈ A
@test norm(lu_factor.P*L*U - A) < 1e-10 

#= 
补充：
  1、按矩阵结构分解
    分解类型       适用矩阵       分解形式       核心优势               典型场景
    Cholesky |  对称正定矩阵   |  A=LL'  | 速度最快、数值最稳定  | 协方差矩阵、二次型优化
    QR       |  任意满秩矩阵   |  A=QR   | 数值稳定、适用范围广  | 最小二乘、特征值计算
    LU       |  任意非奇异方阵 |  PA=LU  | 通用性强、计算高效    | 一般线性方程组求解
  
  2、正定的含义
     random positive definite matrix 随机正定矩阵
     定义：对于任意非零向量x，二次型x'Ax > 0恒成立
     1.random 随机的
     2.positive 对应">0"：二次型结果始终为正数
     3.definite 对应恒成立，无论选哪个非零向量，结果都是正数
       definite = 确定的、必然的(数学上表述"无例外")
       positive definite = 必然为正的 -> 浓缩为正定

=#
```

```julia
# I can solve this linear system with these factors 
# 我可以用这些分解因子来求解这个线性方程组
b = randn(3)

# solve x = A\b using our factors 
# 用这些分解因子求解 x = A\b
x1 = chol_factor\b 
x2 = qr_factor\b 
x3 = lu_factor\b 

@test norm(x1 - x2) < 1e-10 
@test norm(x1 - x3) < 1e-10 
```

```julia
# this is useful for 2 reasons:
# 这样做有两个好处：
# 1. it helps us avoid bugs by specifying the types we expect 
# 1. 通过明确期望的类型来避免 bug
# 2. it allows for multiple dispatch(多重派发)

# here's what I mean by multiple dispatch
# 下面演示一下什么是多重派发
function print_my_type(a::Float64)
    println("thanks for inputting a ~float~")
end
print_my_type(3.4)
```

```julia
# but if I give it an integer, it won't be able to find the function 
# 但如果传入整数，它就找不到对应的函数了
print_my_type(3)
```

```julia
# this error is saying it doesn't have a method for `print_my_type(::Int64)`
# 报错大意：没有 print_my_type(::Int64) 这个方法
# but the closest it found was `print_my_type(::Float64)`
# 它找到的最接近的是 print_my_type(::Float64)

# so let's write a new one 
# 那我们再写一个新的
function print_my_type(a::Int64)
    println("thanks for inputting an ~integer~")
end
print_my_type(3)

# these two functions have the same name, so the version that gets 
# called is determined by the type of the input 
# 两个函数同名，调用哪个版本由输入类型决定
```

```julia
#同一个函数名，不同输入类型对应不同版本的函数
# 版本1：处理 Float64 类型
function print_my_type(a::Float64)
    println("thanks for inputting a ~float~")
end

# 版本2：处理 Int64 类型
function print_my_type(a::Int64)
    println("thanks for inputting an ~integer~")
end

# 版本3：处理 Vector 类型
function print_my_type(a::Vector)
    println("thanks for inputting a ~vector~")
end

# 调用不同版本
print_my_type(3.4)   # 匹配版本1 → 输出：thanks for inputting a ~float~
print_my_type(3)     # 匹配版本2 → 输出：thanks for inputting an ~integer~
print_my_type([1,2]) # 匹配版本3 → 输出：thanks for inputting a ~vector~
```

```julia
#= 
   to avoid "polluting(污染)" our workspace with global variables, 
   we often times wrap things in(将...包装在...中) a "let end"(let代码块，用于创建一个新的局部作用域) 
=#
function test_local_scope()
     
    # since there are no inputs to this function, it will only know about(访问) global variables
      
    println("here is global_var: ", global_var)
    println("here is local_var: ", local_var) # this will fail 
end

# global variable 
# 全局变量
const global_var = 23

let 
    
    # local variable 
    # 局部变量
    local_var = 45 
    
    test_local_scope()
    
end
```

```julia
# here is an example of global
# 这是一个全局变量的例子
x = 3.4 
function f1()
    if @isdefined(x)
        println("yeah I know about x")
        println("x: ", x)
    end
end
f1()
```

```julia
# functions can modify their inputs in place
# 函数可以原地修改其输入参数
# this is not true in MATLAB, but is true in python 
# 这在MATLAB中不成立，但在Python中成立。

# mutating_function: 变异函数或修改(输入)参数的函数 
function mutating_function!(input::Vector)    
    input[2] = 140
    return nothing 
end

let 
    input = [5,9]
    println("input before: ", input)
    
    # call our function which modifies the input 
    # 调用我们这个会修改输入参数的函数
    mutating_function!(input)
    
    println("input after: ", input)
end
```

```julia
# functions that modify the input have a ! at the end of the title 
# 修改输入的函数在名称末尾带有一个感叹号!

x = [4, 1, 3, 2]
println("here is x: x = ", x)
println("here is the sorted version of x: ", sort(x))  # this returns a new vector, it doesn't modify the input 返回新向量，不修改原输入
println("but x is unchanged: x = ", x) 

# now I will sort the vector in place 
# 现在原地排序这个向量
sort!(x) # ! means it will modify the input 表示会修改输入
println("here is the modified x (it's sorted now): x = ", x)
```

```julia
# plotting 
# 绘图

using Plots 

# our first plot is just a normal one 
# 我们绘制的第一张图就是普通的图

# creating a vector of numbers from -10 to 10, with a step size of .001
# 创建一个从 -10 到 10、步长为 0.001 的数值向量
x = -10:.001:10   

# now we plot the sin of x 
# 现在我们绘制x的正弦函数图像
plot(x, sin.(x), label = "sin(x)", xlabel = "x label",
     ylabel = "y label", xlim = (-5,5), ylim = (-2, 2))
#=
     1. x:第一个参数，指定 x 轴的数据（即上面生成的范围数组）。
     2. sin.(x):第二个参数，指定 y 轴的数据，这里使用了广播操作符（.）将 sin 函数应用到 x 的每个元素上，得到对应的 y 值。
     3. label = "sin(x)":为曲线添加标签，图例中显示
     4. xlabel = "x label":设置 x 轴的标签。
     5. ylabel = "y label":设置 y 轴的标签。
     6. xlim = (-5,5):设置 x 轴的范围，这里是从 -5 到 5。
     7. ylim = (-2, 2):设置 y 轴的范围，这里是从 -2 到 2.
=# 
```

```julia
# now let's plot multiple things on the same plot
# 现在我们在同一张图中绘制多个图形。
plot(x, sin.(x), label = "sin(x)", xlabel = "x label",
     ylabel = "y label", xlim = (-5,5), ylim = (-2, 2))

# now I use plot! to modify my existing plot
# 现在我使用 plot! 来修改已有的图表。
plot!(x, cos.(x), label = "cos(x)") # plot! 在原图上绘制第二条曲线
scatter!(randn(100), randn(100), label = "random scatter points") # scatter! 
                                                                  # 在原图上绘制散点图
```

```julia
# now let's plot a trajectory that's stored as vector of vectors 
# 现在我们来绘制一条以**向量的向量**形式存储的轨迹。

time_vec = 0:.01:10 # 生成时间向量，从0到10，步长为0.01

# create a trajectory as a vector of vectors 
# 创建一个由向量构成的轨迹数据
X = [[exp(-t), exp(-.5*t), exp(-.25*t)] for t in time_vec]

# convert this vector of vectors to a matrix,
# 把“向量的向量”转换成矩阵
#= 
1.这里是由于vector of vectors是"指针数组"，分散存储的，而矩阵是连续的数据块，矩阵更利于绘图
 
2.矩阵：支持高效的广播。例如 Y * 2，整个矩阵的所有元素瞬间乘以 2。
  向量的向量：不支持直接的数学运算。你不能写 X * 2(如果 X 是向量的向量),必须写 [x * 2 for x in X]。
  这会阻碍绘图库进行快速的数学预处理。
=#
Xmatrix = hcat(X...)

# plot this (DONT FORGET THE TRANSPOSE)
# 绘图（千万别忘了转置）
#= 
1. time_vec:指定 x 轴数据，这里是时间向量
2. Xmatrix':指定 y 轴数据，Xmatrix' 是轨迹矩阵的转置
3. label = ["exp(-t)" "exp(-.5t)" "exp(-.25t)"]:为三条轨迹分别添加标签
4. ylabel = "ylabel":设置 y 轴的标签
5. xlabel = "time (s)":设置 x 轴的标签
=# 
plot(time_vec, Xmatrix', label = ["exp(-t)" "exp(-.5t)" "exp(-.25t)"],
     ylabel = "ylabel", xlabel = "time (s)")
  
#=
Xmatrix' (转置操作符)：单引号 ' 表示矩阵转置。
     为什么需要转置？
     Xmatrix 目前的形状是 3×1001（3行，1001列）。
     plot 函数通常期望：每一列代表一条独立的曲线（一个变量），每一行代表一个时间点。
     因此，我们需要 1001×3 的矩阵（1001行，3列）。
     转置后：第 1 列变成了 𝑒−𝑡，第 2 列变成了 𝑒−0.5𝑡的数据，第 3 列变成了 𝑒−0.25𝑡的数据。
     label = ["exp(-t)" "exp(-.5t)" "exp(-.25t)"]：
     这里使用了一个字符串数组（注意空格分隔，不是逗号）。
     因为 plot 发现 Xmatrix' 有 3 列，它会自动将这 3 个标签分别对应到这 3 条曲线上。
=#
```

```julia
## 导入 ForwardDiff 库并简写为 FD，用于自动微分（计算导数）
import ForwardDiff as FD

# 定义一个函数，用于计算函数 f 在点 x̄ 处的一阶泰勒展开（即线性化/切线）在 x 处的值
# f::Function: 目标函数
# x̄::Float64: 展开/线性化的点（切点）
# x::Float64: 需要计算近似值的点
# ::Float64: 指定返回值类型为浮点数
function first_order_taylor(f::Function, x̄::Float64, x::Float64)::Float64
    # 一阶泰勒级数公式：f(x̄) + f'(x̄) * (x - x̄)
    # FD.derivative(f, x̄) 自动计算 f 在 x̄ 处的导数（斜率）
    return f(x̄) + FD.derivative(f,x̄)*(x - x̄)     # alt键+120 x̄
end

let 
    # 1. 生成 x 轴数据：从 -6 到 6，步长 0.01
    x = -6:.01:6
    
    # 2. 生成 y 轴数据：计算 x 对应的正弦值（真实曲线）
    y = sin.(x)
    
    # 3. 随机选择一个线性化点（切点）
    # randn() 生成一个标准正态分布的随机数
    x̄ = randn()

    # 4. 计算泰勒展开的近似值（切线上的 y 值）
    # 对 x 中的每一个点，应用 first_order_taylor 公式
    y_taylor = [first_order_taylor(sin, x̄, x[i]) for i = 1:length(x)]
    # 注：也可以写成更简洁的广播形式: y_taylor = [first_order_taylor(sin, x̄, xi) for xi in x] 
    
    # 5. 绘图部分
    # 绘制随机选中的切点 (x̄, sin(x̄))
    scatter([x̄], [sin(x̄)], label = "linearization point")
    
    # 绘制一阶泰勒展开（切线）
    plot!(x, y_taylor, label = "1st order taylor series")
    
    # 绘制真实的 sin(x) 曲线，并显示图像
    # plot! 表示在当前图上叠加，而不是新建
    display(plot!(x,y, label = "sin(x)", xlabel = "x", ylabel = "y"))
    
end
```
