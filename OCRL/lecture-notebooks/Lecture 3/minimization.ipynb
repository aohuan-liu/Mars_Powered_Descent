```julia
# ========================================
#  环境初始化：激活项目环境
# ========================================
import Pkg; Pkg.activate(@__DIR__); Pkg.instantiate()
```

```julia
# ========================================
#  导入依赖：LinearAlgebra, ForwardDiff, PyPlot
# ========================================
using LinearAlgebra
using ForwardDiff
using PyPlot
```

```julia
# ========================================
#  目标函数 f(x)：非凸函数
#  包含局部最小值、局部最大值和全局最小值
# ========================================
function f(x)
    return x.^4 + x.^3 - x.^2 - x
end
```

```julia
# ========================================
#  梯度 ∂f(x)：用 ForwardDiff 自动微分计算
# ========================================
function ∇f(x)
    return 4.0*x.^3 + 3.0*x.^2 - 2.0*x - 1.0
end
```

```julia
# ========================================
#  Hessian 矩阵 ∂²f(x)：用 ForwardDiff 自动微分计算
# ========================================
function ∇2f(x)
    return 12.0*x.^2 + 6.0*x - 2.0
end
```

```julia
# ========================================
#  描点画出目标函数 f(x) 曲线
# ========================================
x = LinRange(-1.75,1.25,1000)
```

```julia
# ========================================
#  绘制 f(x) 图像
# ========================================
p = plot(x,f(x))
```

```julia
# ========================================
#  Newton 法步长一步
#  x_{new} = x - [∂²f(x)]⁻¹ ∂f(x)
# ========================================
function newton_step(x0)
    xn = x0 - ∇2f(x0)\∇f(x0)
end
```

```julia
# ========================================
#  初始猜测值
# ========================================
xguess = 0.0
plot(x, f(x))
plot(xguess, f(xguess), "rx")
```

```julia
# ========================================
#  迭代 Newton 步
# ========================================
xnew = newton_step(xguess[end])
xguess = [xguess xnew]
plot(x, f(x))
plot(xguess, f(xguess), "rx")
```

```julia
# ========================================
#  检查 Hessian 在 x=0 处是否正定
# ========================================
∇2f(0.0)
```

```julia
# ========================================
#  正则化 Newton 法
#  当 Hessian 非正定时，加对角正则化项确保下降方向
#  ∂²f → ∂²f + βI  (β 为正则化系数)
# ========================================
function regularized_newton_step(x0)
    β = 1.0
    H = ∇2f(x0)
    while !isposdef(H)
        H = H + β*I
    end
    xn = x0 - H\∇f(x0)
end
```

```julia
# ========================================
#  初始猜测值
# ========================================
xguess = 0.0
plot(x, f(x))
plot(xguess, f(xguess), "rx")
```

```julia
# ========================================
#  正则化 Newton 迭代
# ========================================
xnew = regularized_newton_step(xguess[end])
xguess = [xguess xnew]
plot(x, f(x))
plot(xguess, f(xguess), "rx")
```
